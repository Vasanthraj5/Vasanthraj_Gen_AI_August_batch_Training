```python
from google import genai
from google.genai import types
from google.colab import userdata

client = genai.Client(api_key=userdata.get('GEMINI_API_KEY'))
MODEL = 'gemini-3.5-flash-lite'
```

```python
def generate_story(topic, genre, length_word=200):
  prompt = f"""
    Write a creative story based on the following requirements:

    Topic: {topic}
    Genre: {genre}
    Approximate length: {length_word} words

    Requirements:
    - Give the story a suitable title.
    - Make the story engaging and creative.
    - Follow the selected genre.
    - Keep the story close to the requested word count.
    - Write only the story and title.
    """
  return prompt
```

```python
topic = input("Enter the story topic: ")
genre = input("Enter the genre: ")
length_word = int(input("Enter the desired word count: "))

# The 'generate_story' function (defined in cell eLFVawcWvSf6)
# should be updated to return the 'prompt' string it generates.
# For example, add 'return prompt' at the end of 'generate_story' in cell eLFVawcWvSf6.
story_prompt = generate_story(topic, genre, length_word)

# Generate the story content using the model
response = client.models.generate_content(
        model=MODEL,
        contents = story_prompt
    )

story = response.text

# Display the story

print("GENERATED STORY")

print(story)
```
