```python
from google import genai
from google.genai import types
from google.colab import userdata

client = genai.Client(api_key=userdata.get('GEMINI_API_KEY'))
MODEL = 'gemini-3.5-flash-lite'
```

```python
knowledge_base = """
Q: What are your office timings?
A: Our office is open from 9:00 AM to 5:00 PM, Monday to Friday.

Q: Is the office open on Saturday?
A: The office is open from 9:00 AM to 1:00 PM on working Saturdays.

Q: How can I apply for admission?
A: You can submit the required application form and documents at the department office.

Q: What documents are required for admission?
A: Students must submit previous academic certificates, ID proof, passport-size photographs, and other documents requested by the office.

Q: What is the attendance requirement?
A: Students must maintain at least 75 percent attendance to meet the minimum attendance requirement.

Q: How can I check my attendance?
A: You can check your attendance through the college portal or contact the department office.

Q: What are the department office working hours during lunch?
A: The lunch break is from 1:00 PM to 2:00 PM.

Q: How can I get my bonafide certificate?
A: Submit a request at the department office and collect the certificate after it is processed.

Q: How long does it take to get a bonafide certificate?
A: The bonafide certificate is normally processed within 2 working days.

Q: How can I apply for a leave?
A: Submit a leave request to your class advisor or department office with the required reason.

Q: What should I do if I lose my college ID card?
A: Report the loss to the department office and follow the procedure for applying for a replacement ID card.

Q: How much is the replacement ID card fee?
A: The replacement ID card fee is Rs. 200.

Q: How can I know the exam timetable?
A: The exam timetable will be announced through the college portal or department notice board.

Q: Where can I submit my assignment?
A: Submit your assignment to the faculty member or through the platform specified by your department.

Q: What are the computer lab timings?
A: The computer lab is available from 9:00 AM to 4:30 PM on working days.

Q: Can students use the lab after class hours?
A: Students can use the lab only with permission from the concerned faculty member.

Q: How can I contact the department office?
A: You can visit the department office during working hours or contact the office through the official department contact details.

Q: Ok
A: Thanks for using the CSE Department Support Chatbot!
"""


system_instruction = f"""
You are a helpful support chatbot for the CSE Department Office.

Your job is to answer student questions using ONLY the information
provided in the knowledge base below.

RULES:

1. Answer only from the knowledge base.
2. Do not invent information.
3. Do not use outside knowledge.
4. If the answer is not available in the knowledge base, reply with
   exactly this sentence:

   "Sorry, I don't have information about that. Please contact the CSE Department Office at support@example.com."

5. Keep answers clear and helpful.
6. Remember the conversation history and use previous messages when relevant.

KNOWLEDGE BASE:

{knowledge_base}
"""

```

```python
chat = client.chats.create(
    model=MODEL,
    config = types.GenerateContentConfig(
        system_instruction=system_instruction,
    ),
    history=[]
)

print("Chatbot is ready!")

print("CSE Department Support Chatbot\n")
print("type 'quit' to exit")

while True:
  user_input = input("You : ")
  if user_input.lower() in ["quit", 'bye', "exit"]:
    print("\nBot: Thank you for using the CSE Department Support Chatbot!")
    break


  reply = chat.send_message(user_input)
  print(f"Bot: {reply.text}\n")
```
